```python
# =============================================================================
# NEUROVISION AI
# FULL FIXED SINGLE-CELL KAGGLE PIPELINE
#
# 3-CLASS BRAIN TUMOR MRI CLASSIFICATION
# EfficientNet-B0 + Patient-Level Split + 50 Epochs + Early Stopping + XAI
#
# FIXES:
#   1. HDF5 MATLAB v7.3 loading
#   2. cvind.mat HDF5 loading
#   3. Patient-level leakage prevention
#   4. Grayscale -> RGB conversion
#   5. ImageNet normalization channel mismatch
#   6. DataLoader storage issue
#   7. T4 CUDA compatibility
#   8. 50-epoch training
#   9. Early stopping
#  10. Robust multiclass ROC-AUC
#  11. Non-blank ROC curves
#  12. ROC-AUC history plotting
#  13. PR-AUC
#  14. Full classification metrics
#  15. Confusion matrix
#  16. Grad-CAM
#  17. XAI tumor-mask localization
#  18. Training history
#  19. Model parameter history
#  20. Deployment export
# =============================================================================


# =============================================================================
# 00 | IMPORTS
# =============================================================================

import os
import sys
import json
import time
import math
import random
import platform
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import h5py

from tqdm.auto import tqdm

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score,
    matthews_corrcoef,
    cohen_kappa_score,
    roc_curve,
    auc,
)

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader

from torchvision import models, transforms

warnings.filterwarnings("ignore")


# =============================================================================
# 01 | PROJECT CONFIGURATION
# =============================================================================

PROJECT_NAME = "NeuroVision AI"

SEED = 42

DATA_ROOT = Path(
    "/kaggle/input/datasets/sudipde25/"
    "mri-dataset-for-detection-and-analysis"
)

MAT_ROOT = (
    DATA_ROOT /
    "BrainTumorDataPublic"
)

CVIND_PATH = (
    DATA_ROOT /
    "cvind.mat"
)

OUTPUT_ROOT = (
    Path("/kaggle/working/neurovision")
)

MANIFEST_DIR = OUTPUT_ROOT / "manifest"
CHECKPOINT_DIR = OUTPUT_ROOT / "checkpoints"
METRICS_DIR = OUTPUT_ROOT / "metrics"
FIGURES_DIR = OUTPUT_ROOT / "figures"
XAI_DIR = OUTPUT_ROOT / "xai"
EXPORT_DIR = OUTPUT_ROOT / "exports"
LOG_DIR = OUTPUT_ROOT / "logs"


for directory in [
    OUTPUT_ROOT,
    MANIFEST_DIR,
    CHECKPOINT_DIR,
    METRICS_DIR,
    FIGURES_DIR,
    XAI_DIR,
    EXPORT_DIR,
    LOG_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# =============================================================================
# 02 | TRAINING CONFIGURATION
# =============================================================================

IMAGE_SIZE = 224

BATCH_SIZE = 16

# REQUESTED
EPOCHS = 50

PATIENCE = 7

LEARNING_RATE = 1e-4

WEIGHT_DECAY = 1e-4

LABEL_SMOOTHING = 0.05

GRAD_CLIP = 1.0

NUM_CLASSES = 3

TEST_FOLD = 5

VAL_SIZE = 0.20

NUM_WORKERS = 0

CLASS_NAMES = [
    "Meningioma",
    "Glioma",
    "Pituitary",
]

LABEL_MAP = {
    1: "Meningioma",
    2: "Glioma",
    3: "Pituitary",
}


# =============================================================================
# 03 | REPRODUCIBILITY
# =============================================================================

def seed_everything(seed=42):

    random.seed(seed)

    np.random.seed(seed)

    os.environ["PYTHONHASHSEED"] = str(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed(seed)

        torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.benchmark = True

    torch.backends.cudnn.deterministic = False


seed_everything(SEED)


# =============================================================================
# 04 | LOGGING
# =============================================================================

LOG_FILE = (
    LOG_DIR /
    f"neurovision_"
    f"{datetime.now().strftime('%Y%m%d_%H%M%S')}"
    f".log"
)


def log(message="", level="INFO"):

    timestamp = datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    text = (
        f"[{timestamp}] "
        f"[{level}] "
        f"{message}"
    )

    print(text)

    with open(
        LOG_FILE,
        "a",
        encoding="utf-8"
    ) as file:

        file.write(
            text + "\n"
        )


def section(title):

    log()

    log("=" * 100)

    log(title)

    log("=" * 100)


# =============================================================================
# 05 | HARDWARE
# =============================================================================

section(
    "STEP 01 | HARDWARE VALIDATION"
)

DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

USE_AMP = (
    DEVICE.type == "cuda"
)

PIN_MEMORY = (
    DEVICE.type == "cuda"
)


log(f"Device: {DEVICE}")

log(f"PyTorch: {torch.__version__}")

log(f"Python: {platform.python_version()}")

log(f"CUDA: {torch.version.cuda}")


if DEVICE.type == "cuda":

    GPU_NAME = torch.cuda.get_device_name(0)

    GPU_MEMORY = (
        torch.cuda.get_device_properties(0)
        .total_memory /
        (1024 ** 3)
    )

    CAPABILITY = (
        torch.cuda.get_device_capability(0)
    )

    log(f"GPU: {GPU_NAME}")

    log(
        f"GPU Memory: "
        f"{GPU_MEMORY:.2f} GB"
    )

    log(
        f"Compute Capability: "
        f"{CAPABILITY[0]}.{CAPABILITY[1]}"
    )

    log(
        "Architecture: "
        f"sm_{CAPABILITY[0]}"
        f"{CAPABILITY[1]}"
    )

    try:

        test_conv = nn.Conv2d(
            3,
            8,
            kernel_size=3,
            padding=1
        ).to(DEVICE)

        test_input = torch.randn(
            2,
            3,
            64,
            64,
            device=DEVICE
        )

        with torch.no_grad():

            test_output = test_conv(
                test_input
            )

        torch.cuda.synchronize()

        del test_conv
        del test_input
        del test_output

        torch.cuda.empty_cache()

        log(
            "CUDA convolution test PASSED"
        )

    except Exception as error:

        raise RuntimeError(
            "CUDA validation failed:\n"
            f"{error}"
        )

else:

    GPU_NAME = "CPU"

    log(
        "CUDA unavailable",
        "WARNING"
    )


# =============================================================================
# 06 | DATASET DISCOVERY
# =============================================================================

section(
    "STEP 02 | DATASET DISCOVERY"
)

if not DATA_ROOT.exists():

    raise FileNotFoundError(
        f"Dataset root not found:\n"
        f"{DATA_ROOT}"
    )

if not MAT_ROOT.exists():

    raise FileNotFoundError(
        f"BrainTumorDataPublic not found:\n"
        f"{MAT_ROOT}"
    )


MAT_FILES = sorted(
    MAT_ROOT.glob("*.mat"),
    key=lambda path:
        int(path.stem)
        if path.stem.isdigit()
        else path.stem
)


log(
    f"BrainTumorDataPublic MAT files: "
    f"{len(MAT_FILES):,}"
)


if len(MAT_FILES) == 0:

    raise RuntimeError(
        "No MAT files found."
    )


# =============================================================================
# 07 | HDF5 MAT READER
# =============================================================================

def decode_pid(pid_array):

    arr = np.asarray(
        pid_array
    ).flatten()

    chars = []

    for value in arr:

        try:

            value = int(value)

            if 0 < value < 256:

                chars.append(
                    chr(value)
                )

        except Exception:

            pass

    pid_string = "".join(chars)

    if pid_string == "":

        return "UNKNOWN"

    if pid_string.isdigit():

        return int(pid_string)

    return pid_string


def read_brain_mat(path):

    with h5py.File(
        str(path),
        "r"
    ) as file:

        cjdata = file["cjdata"]

        image = np.array(
            cjdata["image"]
        )

        label = int(
            np.array(
                cjdata["label"]
            ).squeeze()
        )

        pid = decode_pid(
            np.array(
                cjdata["PID"]
            )
        )

        mask = np.array(
            cjdata["tumorMask"]
        )

        border = np.array(
            cjdata["tumorBorder"]
        )

    return (
        image,
        label,
        pid,
        mask,
        border
    )


# =============================================================================
# 08 | MAT FORMAT VALIDATION
# =============================================================================

section(
    "STEP 03 | MAT FILE FORMAT VALIDATION"
)

test_file = MAT_FILES[0]

log(
    f"Test file: {test_file.name}"
)

(
    test_image,
    test_label,
    test_pid,
    test_mask,
    test_border
) = read_brain_mat(
    test_file
)

log(
    f"Image shape: {test_image.shape}"
)

log(
    f"Label: {test_label}"
)

log(
    f"Class: "
    f"{LABEL_MAP.get(test_label, 'Unknown')}"
)

log(
    f"PID: {test_pid}"
)

log(
    f"Mask shape: {test_mask.shape}"
)

if test_label not in [1, 2, 3]:

    raise ValueError(
        f"Invalid label: {test_label}"
    )

log(
    "BrainTumorDataPublic HDF5 validation PASSED"
)


# =============================================================================
# 09 | CV INDEX
# =============================================================================

section(
    "STEP 04 | CROSS-VALIDATION INDEX"
)

if not CVIND_PATH.exists():

    raise FileNotFoundError(
        f"cvind.mat not found:\n"
        f"{CVIND_PATH}"
    )


with h5py.File(
    str(CVIND_PATH),
    "r"
) as file:

    log(
        f"cvind root keys: "
        f"{list(file.keys())}"
    )

    if "cvind" not in file:

        raise KeyError(
            "cvind dataset missing"
        )

    cvind = np.array(
        file["cvind"]
    ).squeeze()


cvind = (
    cvind
    .astype(int)
    .flatten()
)


log(
    f"cvind samples: "
    f"{len(cvind):,}"
)

log(
    f"Unique folds: "
    f"{sorted(np.unique(cvind).tolist())}"
)


if len(cvind) != len(MAT_FILES):

    raise ValueError(
        "cvind length does not match MAT file count."
    )


if set(np.unique(cvind)) != {
    1, 2, 3, 4, 5
}:

    raise ValueError(
        "Unexpected fold values."
    )


log(
    "CV index validation PASSED"
)


# =============================================================================
# 10 | BUILD MANIFEST
# =============================================================================

section(
    "STEP 05 | BUILDING DATASET MANIFEST"
)

records = []

failed_files = []


for index, path in enumerate(
    tqdm(
        MAT_FILES,
        desc="Reading BrainTumorDataPublic"
    )
):

    try:

        (
            image,
            label,
            pid,
            mask,
            border
        ) = read_brain_mat(
            path
        )

        tumor_pixels = int(
            np.count_nonzero(mask)
        )

        records.append({

            "sample_id":
                index + 1,

            "file":
                str(path),

            "pid":
                pid,

            "label":
                int(label),

            "class_name":
                LABEL_MAP[int(label)],

            "cv_fold":
                int(cvind[index]),

            "image_height":
                int(image.shape[0]),

            "image_width":
                int(image.shape[1]),

            "mask_height":
                int(mask.shape[0]),

            "mask_width":
                int(mask.shape[1]),

            "tumor_pixels":
                tumor_pixels,

            "has_tumor_mask":
                tumor_pixels > 0,

        })

    except Exception as error:

        failed_files.append({

            "file":
                path.name,

            "error":
                str(error)

        })


log(
    f"Successfully extracted: "
    f"{len(records):,}"
)

log(
    f"Failed files: "
    f"{len(failed_files):,}"
)


if failed_files:

    failed_df = pd.DataFrame(
        failed_files
    )

    failed_df.to_csv(
        MANIFEST_DIR /
        "failed_files.csv",
        index=False
    )

    raise RuntimeError(
        "Some MAT files failed."
    )


manifest = pd.DataFrame(records)


MANIFEST_PATH = (
    MANIFEST_DIR /
    "brain_tumor_manifest.csv"
)


manifest.to_csv(
    MANIFEST_PATH,
    index=False
)


log(
    f"Manifest saved: "
    f"{MANIFEST_PATH}"
)


# =============================================================================
# 11 | DATASET SUMMARY
# =============================================================================

section(
    "STEP 06 | DATASET SUMMARY"
)

log(
    f"Total samples: "
    f"{len(manifest):,}"
)

log(
    f"Unique patients: "
    f"{manifest['pid'].nunique():,}"
)

log(
    f"Classes: "
    f"{manifest['label'].nunique()}"
)


label_summary = (
    manifest
    .groupby(
        [
            "label",
            "class_name"
        ]
    )
    .agg(

        samples=(
            "sample_id",
            "count"
        ),

        patients=(
            "pid",
            "nunique"
        ),

        mean_tumor_pixels=(
            "tumor_pixels",
            "mean"
        )

    )
    .reset_index()
)


print(
    "\nLABEL DISTRIBUTION\n"
)

print(
    label_summary.to_string(
        index=False
    )
)


# =============================================================================
# 12 | PATIENT LABEL CONSISTENCY
# =============================================================================

section(
    "STEP 07 | PATIENT LABEL CONSISTENCY"
)

patient_label_counts = (
    manifest
    .groupby("pid")["label"]
    .nunique()
)


single_label_patients = int(
    (patient_label_counts == 1).sum()
)

multi_label_patients = int(
    (patient_label_counts > 1).sum()
)


log(
    f"Patients with exactly one label: "
    f"{single_label_patients}"
)

log(
    f"Patients with multiple labels: "
    f"{multi_label_patients}"
)


if multi_label_patients > 0:

    raise ValueError(
        "Patient label inconsistency."
    )


log(
    "PATIENT LABEL CONSISTENCY PASSED"
)


# =============================================================================
# 13 | PATIENT FOLD CONSISTENCY
# =============================================================================

section(
    "STEP 08 | PATIENT FOLD CONSISTENCY"
)

patient_fold_counts = (
    manifest
    .groupby("pid")["cv_fold"]
    .nunique()
)


single_fold_patients = int(
    (patient_fold_counts == 1).sum()
)

multi_fold_patients = int(
    (patient_fold_counts > 1).sum()
)


log(
    f"Patients in exactly one fold: "
    f"{single_fold_patients}"
)

log(
    f"Patients in multiple folds: "
    f"{multi_fold_patients}"
)


if multi_fold_patients > 0:

    raise ValueError(
        "Patient appears in multiple folds."
    )


log(
    "PATIENT FOLD CONSISTENCY PASSED"
)


# =============================================================================
# 14 | FOLD SUMMARY
# =============================================================================

fold_summary = (
    manifest
    .groupby(
        [
            "cv_fold",
            "class_name"
        ]
    )
    .size()
    .unstack(fill_value=0)
)


print(
    "\nCV FOLD × CLASS\n"
)

print(
    fold_summary
)


# =============================================================================
# 15 | PATIENT-LEVEL TRAIN / VALIDATION / TEST
# =============================================================================

section(
    "STEP 09 | PATIENT-LEVEL TRAIN / VALIDATION / TEST"
)

test_df = (
    manifest[
        manifest["cv_fold"] == TEST_FOLD
    ]
    .copy()
)


trainval_df = (
    manifest[
        manifest["cv_fold"] != TEST_FOLD
    ]
    .copy()
)


test_patients = set(
    test_df["pid"]
)


patient_table = (
    trainval_df
    .groupby("pid")
    .agg(
        label=(
            "label",
            "first"
        )
    )
    .reset_index()
)


train_patient_ids, val_patient_ids = train_test_split(
    patient_table["pid"].values,
    test_size=VAL_SIZE,
    random_state=SEED,
    stratify=patient_table["label"].values
)


train_patients = set(
    train_patient_ids
)

val_patients = set(
    val_patient_ids
)


train_df = (
    trainval_df[
        trainval_df["pid"].isin(
            train_patients
        )
    ]
    .copy()
)


val_df = (
    trainval_df[
        trainval_df["pid"].isin(
            val_patients
        )
    ]
    .copy()
)


assert train_patients.isdisjoint(
    val_patients
)

assert train_patients.isdisjoint(
    test_patients
)

assert val_patients.isdisjoint(
    test_patients
)


log(
    f"Test fold: {TEST_FOLD}"
)

log(
    f"Train+Validation samples: "
    f"{len(trainval_df):,}"
)

log(
    f"Test samples: "
    f"{len(test_df):,}"
)

log(
    f"Train patients: "
    f"{len(train_patients):,}"
)

log(
    f"Validation patients: "
    f"{len(val_patients):,}"
)

log(
    f"Test patients: "
    f"{len(test_patients):,}"
)

log(
    f"Train samples: "
    f"{len(train_df):,}"
)

log(
    f"Validation samples: "
    f"{len(val_df):,}"
)

log(
    f"Test samples: "
    f"{len(test_df):,}"
)

log(
    f"Train ∩ Validation = "
    f"{len(train_patients & val_patients)}"
)

log(
    f"Train ∩ Test = "
    f"{len(train_patients & test_patients)}"
)

log(
    f"Validation ∩ Test = "
    f"{len(val_patients & test_patients)}"
)

log(
    "PATIENT-LEVEL LEAKAGE CHECK PASSED"
)


train_df.to_csv(
    MANIFEST_DIR /
    "train_manifest.csv",
    index=False
)

val_df.to_csv(
    MANIFEST_DIR /
    "validation_manifest.csv",
    index=False
)

test_df.to_csv(
    MANIFEST_DIR /
    "test_manifest.csv",
    index=False
)


# =============================================================================
# 16 | MRI NORMALIZATION
# =============================================================================

section(
    "STEP 10 | MRI PREPROCESSING"
)


def normalize_mri(image):

    image = np.asarray(
        image,
        dtype=np.float32
    )

    image = np.nan_to_num(
        image,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    low = np.percentile(
        image,
        1
    )

    high = np.percentile(
        image,
        99
    )

    if high > low:

        image = (
            image - low
        ) / (
            high - low
        )

    else:

        image = np.zeros_like(
            image
        )

    image = np.clip(
        image,
        0.0,
        1.0
    )

    image = (
        image * 255
    ).astype(
        np.uint8
    )

    return image


# =============================================================================
# 17 | MRI -> RGB PREPROCESSING
# =============================================================================

IMAGENET_MEAN = [
    0.485,
    0.456,
    0.406
]

IMAGENET_STD = [
    0.229,
    0.224,
    0.225
]


train_transform = transforms.Compose([

    transforms.ToPILImage(),

    transforms.Grayscale(
        num_output_channels=3
    ),

    transforms.Resize(
        (
            IMAGE_SIZE,
            IMAGE_SIZE
        )
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=IMAGENET_MEAN,
        std=IMAGENET_STD
    ),

])


eval_transform = transforms.Compose([

    transforms.ToPILImage(),

    transforms.Grayscale(
        num_output_channels=3
    ),

    transforms.Resize(
        (
            IMAGE_SIZE,
            IMAGE_SIZE
        )
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=IMAGENET_MEAN,
        std=IMAGENET_STD
    ),

])


log(
    "MRI preprocessing: "
    "grayscale -> 3-channel RGB -> "
    "224x224 -> ImageNet normalization"
)


# =============================================================================
# 18 | CLASS BALANCING
# =============================================================================

section(
    "STEP 11 | CLASS BALANCING"
)

train_counts = (
    train_df["label"]
    .value_counts()
    .sort_index()
)


class_weights = (
    len(train_df)
    /
    (
        NUM_CLASSES *
        train_counts.values
    )
)


class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=DEVICE
)


for index, weight in enumerate(
    class_weights
):

    log(
        f"{CLASS_NAMES[index]}: "
        f"{weight.item():.4f}"
    )


# =============================================================================
# 19 | DATASET
# =============================================================================

class BrainTumorDataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform,
        return_mask=False
    ):

        self.df = (
            dataframe
            .reset_index(drop=True)
            .copy()
        )

        self.transform = transform

        self.return_mask = return_mask


    def __len__(self):

        return len(self.df)


    def __getitem__(self, index):

        row = self.df.iloc[index]

        (
            image,
            label,
            pid,
            mask,
            border
        ) = read_brain_mat(
            row["file"]
        )

        image = normalize_mri(
            image
        )

        image = self.transform(
            image
        )

        image = image.contiguous()

        target = torch.tensor(
            int(label) - 1,
            dtype=torch.long
        )

        item = {

            "image":
                image,

            "label":
                target,

            "pid":
                str(pid),

            "sample_id":
                int(row["sample_id"]),

            "path":
                str(row["file"]),

        }

        if self.return_mask:

            mask = np.asarray(
                mask,
                dtype=np.float32
            )

            mask = (
                mask > 0
            ).astype(
                np.float32
            )

            mask = np.ascontiguousarray(
                mask
            )

            item["mask"] = (
                torch.from_numpy(
                    mask.copy()
                ).float()
            )

        return item


# =============================================================================
# 20 | DATA LOADERS
# =============================================================================

section(
    "STEP 10B | ROBUST DATASET LOADER"
)

train_dataset = BrainTumorDataset(
    train_df,
    train_transform,
    return_mask=False
)

val_dataset = BrainTumorDataset(
    val_df,
    eval_transform,
    return_mask=False
)

test_dataset = BrainTumorDataset(
    test_df,
    eval_transform,
    return_mask=True
)


train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=PIN_MEMORY,
    drop_last=False
)


val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=PIN_MEMORY,
    drop_last=False
)


test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=PIN_MEMORY,
    drop_last=False
)


log(
    f"Train batches: "
    f"{len(train_loader):,}"
)

log(
    f"Validation batches: "
    f"{len(val_loader):,}"
)

log(
    f"Test batches: "
    f"{len(test_loader):,}"
)


# =============================================================================
# 21 | DATALOADER SANITY TEST
# =============================================================================

section(
    "STEP 10C | DATALOADER SANITY TEST"
)

train_batch = next(
    iter(train_loader)
)


log(
    f"Training image batch shape: "
    f"{tuple(train_batch['image'].shape)}"
)

log(
    f"Training label batch shape: "
    f"{tuple(train_batch['label'].shape)}"
)


assert train_batch["image"].shape[1] == 3

assert train_batch["image"].shape[2] == IMAGE_SIZE

assert train_batch["image"].shape[3] == IMAGE_SIZE

assert train_batch["label"].ndim == 1


log(
    "Training DataLoader sanity test PASSED"
)


val_batch = next(
    iter(val_loader)
)


log(
    f"Validation image batch shape: "
    f"{tuple(val_batch['image'].shape)}"
)

assert val_batch["image"].shape[1] == 3

log(
    "Validation DataLoader sanity test PASSED"
)


test_batch = next(
    iter(test_loader)
)


log(
    f"Test image batch shape: "
    f"{tuple(test_batch['image'].shape)}"
)

log(
    f"Test mask batch shape: "
    f"{tuple(test_batch['mask'].shape)}"
)

assert test_batch["image"].shape[1] == 3

assert test_batch["mask"].ndim == 3


log(
    "Test/XAI DataLoader sanity test PASSED"
)

log(
    "ALL DATALOADER TESTS PASSED"
)


# =============================================================================
# 22 | MODEL
# =============================================================================

section(
    "STEP 12 | MODEL INITIALIZATION"
)

weights = (
    models.EfficientNet_B0_Weights.DEFAULT
)


model = models.efficientnet_b0(
    weights=weights
)


model.classifier[1] = nn.Linear(
    model.classifier[1].in_features,
    NUM_CLASSES
)


model = model.to(
    DEVICE
)


total_params = sum(
    parameter.numel()
    for parameter in model.parameters()
)


trainable_params = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)


log(
    f"Total parameters: "
    f"{total_params:,}"
)

log(
    f"Trainable parameters: "
    f"{trainable_params:,}"
)


parameter_history = pd.DataFrame([{

    "timestamp":
        datetime.now().isoformat(),

    "model":
        "EfficientNet-B0",

    "total_parameters":
        total_params,

    "trainable_parameters":
        trainable_params,

    "image_size":
        IMAGE_SIZE,

    "batch_size":
        BATCH_SIZE,

    "num_classes":
        NUM_CLASSES,

    "learning_rate":
        LEARNING_RATE,

    "weight_decay":
        WEIGHT_DECAY,

    "epochs":
        EPOCHS,

    "patience":
        PATIENCE,

    "gpu":
        GPU_NAME,

}])


parameter_history.to_csv(
    METRICS_DIR /
    "model_parameter_history.csv",
    index=False
)


# =============================================================================
# 23 | LOSS / OPTIMIZER
# =============================================================================

criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=LABEL_SMOOTHING
)


optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)


scheduler = (
    torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="max",
        factor=0.5,
        patience=2,
        min_lr=1e-7
    )
)


scaler = torch.amp.GradScaler(
    "cuda",
    enabled=USE_AMP
)


# =============================================================================
# 24 | METRICS
# =============================================================================

def calculate_metrics(
    y_true,
    y_pred,
    y_prob=None
):

    metrics = {}

    metrics["accuracy"] = accuracy_score(
        y_true,
        y_pred
    )

    metrics["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )

    metrics["precision_macro"] = (
        precision_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["recall_macro"] = (
        recall_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["sensitivity_macro"] = (
        metrics["recall_macro"]
    )

    metrics["f1_macro"] = (
        f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0
        )
    )

    metrics["f1_weighted"] = (
        f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    metrics["mcc"] = (
        matthews_corrcoef(
            y_true,
            y_pred
        )
    )

    metrics["cohen_kappa"] = (
        cohen_kappa_score(
            y_true,
            y_pred
        )
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES))
    )

    specificity_values = []

    for i in range(NUM_CLASSES):

        tp = cm[i, i]

        fn = (
            cm[i, :].sum()
            - tp
        )

        fp = (
            cm[:, i].sum()
            - tp
        )

        tn = (
            cm.sum()
            - tp
            - fn
            - fp
        )

        specificity = (
            tn /
            (tn + fp)
            if (tn + fp) > 0
            else 0.0
        )

        specificity_values.append(
            specificity
        )

    metrics["specificity_macro"] = float(
        np.mean(
            specificity_values
        )
    )

    # -------------------------------------------------------------------------
    # Robust ROC-AUC
    # -------------------------------------------------------------------------

    metrics["roc_auc_ovr_macro"] = np.nan

    metrics["pr_auc_macro"] = np.nan

    if y_prob is not None:

        y_prob = np.asarray(
            y_prob,
            dtype=np.float64
        )

        if (
            y_prob.ndim == 2
            and
            y_prob.shape[1] == NUM_CLASSES
            and
            len(y_true) == len(y_prob)
            and
            len(np.unique(y_true)) == NUM_CLASSES
        ):

            # Defensive renormalization: softmax probabilities computed
            # under float16 autocast can drift slightly off a row-sum of
            # 1.0 (float16 eps ~= 1e-3), which sklearn's roc_auc_score
            # rejects outright for multiclass "ovr"/"ovo" scoring. We
            # clip negatives/NaNs from numerical noise and rescale each
            # row back onto the probability simplex before scoring.
            y_prob_safe = np.clip(
                np.nan_to_num(y_prob, nan=0.0),
                0.0,
                None
            )

            row_sums = y_prob_safe.sum(
                axis=1,
                keepdims=True
            )

            row_sums[row_sums == 0.0] = 1.0

            y_prob_norm = (
                y_prob_safe / row_sums
            )

            try:

                metrics[
                    "roc_auc_ovr_macro"
                ] = float(
                    roc_auc_score(
                        y_true,
                        y_prob_norm,
                        multi_class="ovr",
                        average="macro"
                    )
                )

            except Exception as error:

                log(
                    f"ROC-AUC calculation warning: "
                    f"{error}",
                    "WARNING"
                )

            try:

                y_onehot = np.eye(
                    NUM_CLASSES,
                    dtype=np.float64
                )[y_true]

                metrics[
                    "pr_auc_macro"
                ] = float(
                    average_precision_score(
                        y_onehot,
                        y_prob_norm,
                        average="macro"
                    )
                )

            except Exception as error:

                log(
                    f"PR-AUC calculation warning: "
                    f"{error}",
                    "WARNING"
                )

    return (
        metrics,
        cm
    )


# =============================================================================
# 25 | TRAIN ONE EPOCH
# =============================================================================

def train_one_epoch():

    model.train()

    running_loss = 0.0

    y_true = []

    y_pred = []

    y_prob = []


    progress = tqdm(
        train_loader,
        desc="Training",
        leave=False
    )


    for batch in progress:

        images = (
            batch["image"]
            .to(
                DEVICE,
                non_blocking=True
            )
        )

        labels = (
            batch["label"]
            .to(
                DEVICE,
                non_blocking=True
            )
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        with torch.autocast(
            device_type=DEVICE.type,
            dtype=torch.float16,
            enabled=USE_AMP
        ):

            outputs = model(
                images
            )

            loss = criterion(
                outputs,
                labels
            )


        scaler.scale(
            loss
        ).backward()


        scaler.unscale_(
            optimizer
        )


        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            GRAD_CLIP
        )


        scaler.step(
            optimizer
        )

        scaler.update()


        # Softmax is computed in float32 (outputs.float()) even though
        # the forward pass ran under float16 autocast. float16 only has
        # ~3 decimal digits of precision, so softmax rows computed in
        # float16 can fail to sum to 1.0 within sklearn's tolerance,
        # which is exactly what caused the ROC-AUC ValueError.
        probabilities = torch.softmax(
            outputs.float(),
            dim=1
        )


        predictions = torch.argmax(
            probabilities,
            dim=1
        )


        running_loss += (
            loss.item()
            * images.size(0)
        )


        y_true.extend(
            labels.detach()
            .cpu()
            .numpy()
        )

        y_pred.extend(
            predictions.detach()
            .cpu()
            .numpy()
        )

        y_prob.extend(
            probabilities.detach()
            .cpu()
            .numpy()
        )


    epoch_loss = (
        running_loss /
        len(train_loader.dataset)
    )


    metrics, cm = calculate_metrics(
        np.asarray(y_true),
        np.asarray(y_pred),
        np.asarray(y_prob)
    )


    metrics["loss"] = epoch_loss


    return (
        metrics,
        cm
    )


# =============================================================================
# 26 | VALIDATION
# =============================================================================

@torch.no_grad()
def validate():

    model.eval()

    running_loss = 0.0

    y_true = []

    y_pred = []

    y_prob = []


    progress = tqdm(
        val_loader,
        desc="Validation",
        leave=False
    )


    for batch in progress:

        images = (
            batch["image"]
            .to(
                DEVICE,
                non_blocking=True
            )
        )

        labels = (
            batch["label"]
            .to(
                DEVICE,
                non_blocking=True
            )
        )


        with torch.autocast(
            device_type=DEVICE.type,
            dtype=torch.float16,
            enabled=USE_AMP
        ):

            outputs = model(
                images
            )

            loss = criterion(
                outputs,
                labels
            )


        probabilities = torch.softmax(
            outputs.float(),
            dim=1
        )

        predictions = torch.argmax(
            probabilities,
            dim=1
        )


        running_loss += (
            loss.item()
            * images.size(0)
        )


        y_true.extend(
            labels.cpu().numpy()
        )

        y_pred.extend(
            predictions.cpu().numpy()
        )

        y_prob.extend(
            probabilities.cpu().numpy()
        )


    epoch_loss = (
        running_loss /
        len(val_loader.dataset)
    )


    metrics, cm = calculate_metrics(
        np.asarray(y_true),
        np.asarray(y_pred),
        np.asarray(y_prob)
    )


    metrics["loss"] = epoch_loss


    return (
        metrics,
        cm
    )


# =============================================================================
# 27 | TRAINING
# =============================================================================

section(
    "STEP 13 | MODEL TRAINING"
)

log(
    f"Configured epochs: {EPOCHS}"
)

log(
    f"Early stopping patience: {PATIENCE}"
)


history = []

best_val_f1 = -np.inf

best_epoch = 0

patience_counter = 0

best_checkpoint = (
    CHECKPOINT_DIR /
    "best_model.pt"
)


for epoch in range(
    1,
    EPOCHS + 1
):

    print(
        "\n" +
        "=" * 90
    )

    print(
        f"EPOCH {epoch}/{EPOCHS}"
    )

    print(
        "=" * 90
    )


    start_time = time.time()


    train_metrics, train_cm = (
        train_one_epoch()
    )


    val_metrics, val_cm = (
        validate()
    )


    scheduler.step(
        val_metrics["f1_macro"]
    )


    current_lr = (
        optimizer
        .param_groups[0]["lr"]
    )


    elapsed = (
        time.time() -
        start_time
    )


    row = {

        "epoch":
            epoch,

        "learning_rate":
            current_lr,

        "epoch_time_sec":
            elapsed,

        "train_loss":
            train_metrics["loss"],

        "train_accuracy":
            train_metrics["accuracy"],

        "train_f1":
            train_metrics["f1_macro"],

        "train_sensitivity":
            train_metrics["sensitivity_macro"],

        "train_specificity":
            train_metrics["specificity_macro"],

        "train_auc":
            train_metrics["roc_auc_ovr_macro"],

        "train_pr_auc":
            train_metrics["pr_auc_macro"],

        "val_loss":
            val_metrics["loss"],

        "val_accuracy":
            val_metrics["accuracy"],

        "val_balanced_accuracy":
            val_metrics["balanced_accuracy"],

        "val_f1":
            val_metrics["f1_macro"],

        "val_sensitivity":
            val_metrics["sensitivity_macro"],

        "val_specificity":
            val_metrics["specificity_macro"],

        "val_mcc":
            val_metrics["mcc"],

        "val_auc":
            val_metrics["roc_auc_ovr_macro"],

        "val_pr_auc":
            val_metrics["pr_auc_macro"],

    }


    history.append(
        row
    )


    history_df = pd.DataFrame(
        history
    )


    history_df.to_csv(
        METRICS_DIR /
        "training_history.csv",
        index=False
    )


    print("\nTRAIN")

    print(
        f"Loss        : "
        f"{train_metrics['loss']:.4f}"
    )

    print(
        f"Accuracy    : "
        f"{train_metrics['accuracy']:.4f}"
    )

    print(
        f"F1 Macro    : "
        f"{train_metrics['f1_macro']:.4f}"
    )

    print(
        f"Sensitivity : "
        f"{train_metrics['sensitivity_macro']:.4f}"
    )

    print(
        f"Specificity : "
        f"{train_metrics['specificity_macro']:.4f}"
    )

    print(
        f"ROC-AUC     : "
        f"{train_metrics['roc_auc_ovr_macro']:.4f}"
    )

    print(
        f"PR-AUC      : "
        f"{train_metrics['pr_auc_macro']:.4f}"
    )


    print("\nVALIDATION")

    print(
        f"Loss        : "
        f"{val_metrics['loss']:.4f}"
    )

    print(
        f"Accuracy    : "
        f"{val_metrics['accuracy']:.4f}"
    )

    print(
        f"Balanced Acc: "
        f"{val_metrics['balanced_accuracy']:.4f}"
    )

    print(
        f"F1 Macro    : "
        f"{val_metrics['f1_macro']:.4f}"
    )

    print(
        f"Sensitivity : "
        f"{val_metrics['sensitivity_macro']:.4f}"
    )

    print(
        f"Specificity : "
        f"{val_metrics['specificity_macro']:.4f}"
    )

    print(
        f"MCC         : "
        f"{val_metrics['mcc']:.4f}"
    )

    print(
        f"ROC-AUC     : "
        f"{val_metrics['roc_auc_ovr_macro']:.4f}"
    )

    print(
        f"PR-AUC      : "
        f"{val_metrics['pr_auc_macro']:.4f}"
    )

    print(
        f"LR          : "
        f"{current_lr:.7f}"
    )

    print(
        f"Time        : "
        f"{elapsed:.1f}s"
    )


    # -------------------------------------------------------------------------
    # BEST MODEL
    # -------------------------------------------------------------------------

    if (
        val_metrics["f1_macro"]
        >
        best_val_f1
    ):

        best_val_f1 = (
            val_metrics["f1_macro"]
        )

        best_epoch = epoch

        patience_counter = 0


        checkpoint = {

            "epoch":
                epoch,

            "model_state_dict":
                model.state_dict(),

            "optimizer_state_dict":
                optimizer.state_dict(),

            "scheduler_state_dict":
                scheduler.state_dict(),

            "best_val_f1":
                best_val_f1,

            "class_names":
                CLASS_NAMES,

            "image_size":
                IMAGE_SIZE,

            "num_classes":
                NUM_CLASSES,

            "model_name":
                "EfficientNet-B0",

            "seed":
                SEED,

            "imagenet_mean":
                IMAGENET_MEAN,

            "imagenet_std":
                IMAGENET_STD,

        }


        torch.save(
            checkpoint,
            best_checkpoint
        )


        log(
            f"NEW BEST MODEL | "
            f"Epoch={epoch} | "
            f"Val F1={best_val_f1:.4f}"
        )

    else:

        patience_counter += 1

        log(
            f"No improvement | "
            f"Patience="
            f"{patience_counter}/{PATIENCE}"
        )


        if patience_counter >= PATIENCE:

            log(
                f"EARLY STOPPING | "
                f"Best epoch={best_epoch}"
            )

            break


# =============================================================================
# 28 | LOAD BEST MODEL
# =============================================================================

section(
    "STEP 14 | BEST CHECKPOINT"
)

if not best_checkpoint.exists():

    raise FileNotFoundError(
        "Best checkpoint was not created."
    )


checkpoint = torch.load(
    best_checkpoint,
    map_location=DEVICE,
    weights_only=False
)


model.load_state_dict(
    checkpoint["model_state_dict"]
)


log(
    f"Best epoch: "
    f"{checkpoint['epoch']}"
)

log(
    f"Best validation F1: "
    f"{checkpoint['best_val_f1']:.4f}"
)


# =============================================================================
# 29 | FINAL TEST
# =============================================================================

section(
    "STEP 15 | FINAL TEST EVALUATION"
)


@torch.no_grad()
def evaluate_test():

    model.eval()

    y_true = []

    y_pred = []

    y_prob = []

    sample_ids = []

    pids = []

    paths = []


    for batch in tqdm(
        test_loader,
        desc="Testing"
    ):

        images = (
            batch["image"]
            .to(
                DEVICE,
                non_blocking=True
            )
        )

        labels = (
            batch["label"]
            .to(
                DEVICE,
                non_blocking=True
            )
        )


        with torch.autocast(
            device_type=DEVICE.type,
            dtype=torch.float16,
            enabled=USE_AMP
        ):

            outputs = model(
                images
            )


        probabilities = torch.softmax(
            outputs.float(),
            dim=1
        )

        predictions = torch.argmax(
            probabilities,
            dim=1
        )


        y_true.extend(
            labels.cpu().numpy()
        )

        y_pred.extend(
            predictions.cpu().numpy()
        )

        y_prob.extend(
            probabilities.cpu().numpy()
        )


        sample_ids.extend(
            batch["sample_id"].tolist()
        )

        pids.extend(
            batch["pid"]
        )

        paths.extend(
            batch["path"]
        )


    return (
        np.asarray(y_true),
        np.asarray(y_pred),
        np.asarray(y_prob),
        sample_ids,
        pids,
        paths
    )


(
    y_test,
    y_test_pred,
    y_test_prob,
    test_sample_ids,
    test_pids,
    test_paths
) = evaluate_test()


test_metrics, test_cm = calculate_metrics(
    y_test,
    y_test_pred,
    y_test_prob
)


print(
    "\n" +
    "=" * 90
)

print(
    "FINAL TEST RESULTS"
)

print(
    "=" * 90
)


for key, value in test_metrics.items():

    if isinstance(
        value,
        (
            float,
            np.floating
        )
    ):

        print(
            f"{key:25s}: "
            f"{value:.4f}"
        )


# =============================================================================
# 30 | CLASSIFICATION REPORT
# =============================================================================

classification_report_text = (
    classification_report(
        y_test,
        y_test_pred,
        target_names=CLASS_NAMES,
        zero_division=0
    )
)


print(
    "\nCLASSIFICATION REPORT\n"
)

print(
    classification_report_text
)


classification_report_dict = (
    classification_report(
        y_test,
        y_test_pred,
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0
    )
)


pd.DataFrame(
    classification_report_dict
).T.to_csv(
    METRICS_DIR /
    "classification_report.csv"
)


# =============================================================================
# 31 | CONFUSION MATRIX
# =============================================================================

section(
    "STEP 16 | CONFUSION MATRIX"
)

plt.figure(
    figsize=(8, 7)
)

plt.imshow(
    test_cm,
    interpolation="nearest",
    cmap="Blues"
)

plt.title(
    "NeuroVision AI | Test Confusion Matrix"
)

plt.colorbar()


tick_marks = np.arange(
    NUM_CLASSES
)


plt.xticks(
    tick_marks,
    CLASS_NAMES,
    rotation=30,
    ha="right"
)

plt.yticks(
    tick_marks,
    CLASS_NAMES
)


threshold = (
    test_cm.max() /
    2.0
)


for i in range(NUM_CLASSES):

    for j in range(NUM_CLASSES):

        plt.text(
            j,
            i,
            str(test_cm[i, j]),
            ha="center",
            va="center",
            color=(
                "white"
                if test_cm[i, j] > threshold
                else "black"
            )
        )


plt.ylabel("Actual")

plt.xlabel("Predicted")

plt.tight_layout()


plt.savefig(
    FIGURES_DIR /
    "confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


np.save(
    METRICS_DIR /
    "test_confusion_matrix.npy",
    test_cm
)


# =============================================================================
# 32 | ROBUST ROC CURVES
# =============================================================================

section(
    "STEP 17 | ROC CURVES"
)


assert y_test_prob.ndim == 2

assert y_test_prob.shape[1] == NUM_CLASSES

assert len(y_test) == len(y_test_prob)


# Defensive renormalization for the same reason as in calculate_metrics:
# float16-autocast softmax rows can drift off a sum of 1.0 by more than
# sklearn's tolerance. This does not change class rankings/AUC in any
# meaningful way -- it only restores the rows to a valid probability
# simplex so roc_auc_score stops raising a ValueError.
y_test_prob_safe = np.clip(
    np.nan_to_num(y_test_prob, nan=0.0),
    0.0,
    None
)

y_test_prob_row_sums = y_test_prob_safe.sum(
    axis=1,
    keepdims=True
)

y_test_prob_row_sums[y_test_prob_row_sums == 0.0] = 1.0

y_test_prob_norm = (
    y_test_prob_safe / y_test_prob_row_sums
)


roc_data = {}


for class_index, class_name in enumerate(
    CLASS_NAMES
):

    binary_true = (
        y_test == class_index
    ).astype(np.int32)


    binary_prob = (
        y_test_prob_norm[:, class_index]
    )


    unique_binary = np.unique(
        binary_true
    )


    if len(unique_binary) < 2:

        log(
            f"ROC unavailable for "
            f"{class_name}: "
            f"only one binary class is present.",
            "WARNING"
        )

        continue


    fpr, tpr, thresholds = roc_curve(
        binary_true,
        binary_prob
    )


    class_auc = auc(
        fpr,
        tpr
    )


    roc_data[class_name] = {

        "fpr":
            fpr,

        "tpr":
            tpr,

        "thresholds":
            thresholds,

        "auc":
            float(class_auc),

    }


    log(
        f"{class_name} ROC-AUC: "
        f"{class_auc:.6f}"
    )


if len(roc_data) == 0:

    raise RuntimeError(
        "No ROC curves could be generated."
    )


try:

    macro_auc = roc_auc_score(
        y_test,
        y_test_prob_norm,
        multi_class="ovr",
        average="macro"
    )

except Exception as error:

    log(
        f"Macro ROC-AUC fallback (mean of per-class AUCs) due to: "
        f"{error}",
        "WARNING"
    )

    macro_auc = float(
        np.mean(
            [
                roc_data[name]["auc"]
                for name in roc_data
            ]
        )
    )


log(
    f"Macro ROC-AUC: "
    f"{macro_auc:.6f}"
)


roc_metrics = {

    "Meningioma":
        roc_data.get(
            "Meningioma",
            {}
        ).get(
            "auc",
            np.nan
        ),

    "Glioma":
        roc_data.get(
            "Glioma",
            {}
        ).get(
            "auc",
            np.nan
        ),

    "Pituitary":
        roc_data.get(
            "Pituitary",
            {}
        ).get(
            "auc",
            np.nan
        ),

    "Macro_AUC":
        float(macro_auc),

}


pd.DataFrame(
    [roc_metrics]
).to_csv(
    METRICS_DIR /
    "roc_auc_results.csv",
    index=False
)


# -------------------------------------------------------------------------
# ROC PLOT
# -------------------------------------------------------------------------

plt.figure(
    figsize=(9, 8)
)


for class_name in CLASS_NAMES:

    if class_name not in roc_data:

        continue


    plt.plot(
        roc_data[class_name]["fpr"],
        roc_data[class_name]["tpr"],
        linewidth=2.5,
        label=(
            f"{class_name} "
            f"(AUC = "
            f"{roc_data[class_name]['auc']:.4f})"
        )
    )


plt.plot(
    [0.0, 1.0],
    [0.0, 1.0],
    linestyle="--",
    linewidth=1.5,
    label="Random classifier"
)


plt.xlim(
    0.0,
    1.0
)

plt.ylim(
    0.0,
    1.05
)


plt.xlabel(
    "False Positive Rate",
    fontsize=12
)

plt.ylabel(
    "True Positive Rate",
    fontsize=12
)

plt.title(
    "NeuroVision AI | One-vs-Rest ROC Curves",
    fontsize=14,
    fontweight="bold"
)


plt.text(
    0.58,
    0.08,
    f"Macro AUC = {macro_auc:.4f}",
    fontsize=11
)


plt.grid(
    alpha=0.25
)


plt.legend(
    loc="lower right"
)


plt.tight_layout()


roc_path = (
    FIGURES_DIR /
    "roc_curves.png"
)


plt.savefig(
    roc_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()


log(
    f"ROC curve saved: "
    f"{roc_path}"
)


# =============================================================================
# 33 | TRAINING HISTORY
# =============================================================================

section(
    "STEP 18 | TRAINING HISTORY"
)

history_df = pd.DataFrame(
    history
)


history_df.to_csv(
    METRICS_DIR /
    "training_history.csv",
    index=False
)


plot_groups = [

    (
        "loss",
        [
            "train_loss",
            "val_loss"
        ],
        "Loss"
    ),

    (
        "accuracy",
        [
            "train_accuracy",
            "val_accuracy"
        ],
        "Accuracy"
    ),

    (
        "f1",
        [
            "train_f1",
            "val_f1"
        ],
        "Macro F1"
    ),

    (
        "auc",
        [
            "train_auc",
            "val_auc"
        ],
        "ROC-AUC"
    ),

    (
        "sensitivity",
        [
            "train_sensitivity",
            "val_sensitivity"
        ],
        "Sensitivity"
    ),

    (
        "specificity",
        [
            "train_specificity",
            "val_specificity"
        ],
        "Specificity"
    ),

]


for filename, columns, title in plot_groups:

    plt.figure(
        figsize=(9, 6)
    )


    plotted_any = False


    for column in columns:

        values = pd.to_numeric(
            history_df[column],
            errors="coerce"
        )


        valid = (
            values.notna()
        )


        if valid.any():

            plt.plot(
                history_df.loc[
                    valid,
                    "epoch"
                ],
                values.loc[
                    valid
                ],
                marker="o",
                linewidth=2,
                label=column
            )

            plotted_any = True


    if not plotted_any:

        plt.close()

        log(
            f"No valid data for "
            f"{title} training plot.",
            "WARNING"
        )

        continue


    plt.xlabel(
        "Epoch"
    )

    plt.ylabel(
        title
    )

    plt.title(
        f"NeuroVision AI | {title}"
    )

    plt.legend()

    plt.grid(
        alpha=0.25
    )

    plt.tight_layout()


    plt.savefig(
        FIGURES_DIR /
        f"training_{filename}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# =============================================================================
# 34 | TEST PREDICTIONS
# =============================================================================

predictions_df = pd.DataFrame({

    "sample_id":
        test_sample_ids,

    "pid":
        test_pids,

    "path":
        test_paths,

    "true_label":
        y_test,

    "true_class":
        [
            CLASS_NAMES[x]
            for x in y_test
        ],

    "predicted_label":
        y_test_pred,

    "predicted_class":
        [
            CLASS_NAMES[x]
            for x in y_test_pred
        ],

    "confidence":
        np.max(
            y_test_prob,
            axis=1
        ),

    "prob_meningioma":
        y_test_prob[:, 0],

    "prob_glioma":
        y_test_prob[:, 1],

    "prob_pituitary":
        y_test_prob[:, 2],

})


predictions_df.to_csv(
    METRICS_DIR /
    "test_predictions.csv",
    index=False
)


# =============================================================================
# 35 | GRAD-CAM
# =============================================================================

section(
    "STEP 19 | GRAD-CAM EXPLAINABILITY"
)


class GradCAM:

    def __init__(
        self,
        model,
        target_layer
    ):

        self.model = model

        self.target_layer = target_layer

        self.activations = None

        self.gradients = None


        self.forward_handle = (
            target_layer.register_forward_hook(
                self.forward_hook
            )
        )


        self.backward_handle = (
            target_layer.register_full_backward_hook(
                self.backward_hook
            )
        )


    def forward_hook(
        self,
        module,
        input,
        output
    ):

        self.activations = output


    def backward_hook(
        self,
        module,
        grad_input,
        grad_output
    ):

        self.gradients = grad_output[0]


    def generate(
        self,
        input_tensor,
        target_class
    ):

        self.model.zero_grad(
            set_to_none=True
        )


        output = self.model(
            input_tensor
        )


        score = output[
            0,
            target_class
        ]


        score.backward()


        gradients = self.gradients

        activations = self.activations


        weights = (
            gradients.mean(
                dim=(2, 3),
                keepdim=True
            )
        )


        cam = (
            weights *
            activations
        ).sum(
            dim=1,
            keepdim=True
        )


        cam = F.relu(cam)


        cam = F.interpolate(
            cam,
            size=(
                IMAGE_SIZE,
                IMAGE_SIZE
            ),
            mode="bilinear",
            align_corners=False
        )


        cam = cam[
            0,
            0
        ]


        cam -= cam.min()


        if cam.max() > 0:

            cam /= cam.max()


        return (
            cam.detach()
            .cpu()
            .numpy()
        )


    def close(self):

        self.forward_handle.remove()

        self.backward_handle.remove()


target_layer = (
    model.features[-1]
)


gradcam = GradCAM(
    model,
    target_layer
)


# =============================================================================
# 36 | XAI METRICS
# =============================================================================

def calculate_xai_metrics(
    cam,
    mask
):

    mask_image = Image.fromarray(
        mask.astype(np.uint8)
    )


    mask_resized = np.array(
        mask_image.resize(
            (
                IMAGE_SIZE,
                IMAGE_SIZE
            ),
            Image.Resampling.NEAREST
        )
    )


    mask_binary = (
        mask_resized > 0
    )


    threshold = np.percentile(
        cam,
        85
    )


    cam_binary = (
        cam >= threshold
    )


    intersection = (
        np.logical_and(
            cam_binary,
            mask_binary
        ).sum()
    )


    union = (
        np.logical_or(
            cam_binary,
            mask_binary
        ).sum()
    )


    cam_area = cam_binary.sum()

    mask_area = mask_binary.sum()


    iou = (
        intersection /
        union
        if union > 0
        else 0.0
    )


    precision = (
        intersection /
        cam_area
        if cam_area > 0
        else 0.0
    )


    recall = (
        intersection /
        mask_area
        if mask_area > 0
        else 0.0
    )


    return {

        "cam_iou":
            float(iou),

        "cam_precision":
            float(precision),

        "cam_recall":
            float(recall),

        "cam_area_ratio":
            float(
                cam_area /
                (
                    IMAGE_SIZE *
                    IMAGE_SIZE
                )
            ),

    }


# =============================================================================
# 37 | GENERATE GRAD-CAM
# =============================================================================

MAX_XAI_SAMPLES = min(
    30,
    len(test_dataset)
)


log(
    f"Generating Grad-CAM for "
    f"{MAX_XAI_SAMPLES} test samples."
)


xai_records = []


model.eval()


for index in tqdm(
    range(MAX_XAI_SAMPLES),
    desc="Grad-CAM"
):

    sample = test_dataset[index]


    image_tensor = (
        sample["image"]
        .unsqueeze(0)
        .to(DEVICE)
    )


    model.zero_grad(
        set_to_none=True
    )


    output = model(
        image_tensor
    )


    probability = torch.softmax(
        output.float(),
        dim=1
    )


    predicted_class = int(
        torch.argmax(
            probability,
            dim=1
        ).item()
    )


    confidence = float(
        probability[
            0,
            predicted_class
        ].item()
    )


    cam = gradcam.generate(
        image_tensor,
        predicted_class
    )


    xai_metrics = (
        calculate_xai_metrics(
            cam,
            sample["mask"].numpy()
        )
    )


    xai_records.append({

        "sample_id":
            sample["sample_id"],

        "pid":
            sample["pid"],

        "true_class":
            CLASS_NAMES[
                int(
                    sample["label"].item()
                )
            ],

        "predicted_class":
            CLASS_NAMES[
                predicted_class
            ],

        "confidence":
            confidence,

        **xai_metrics,

    })


    if index < 12:

        original = (
            sample["image"]
            .permute(1, 2, 0)
            .cpu()
            .numpy()
        )


        original = (
            original *
            np.array(IMAGENET_STD)
            +
            np.array(IMAGENET_MEAN)
        )


        original = np.clip(
            original,
            0,
            1
        )


        mask = sample["mask"].numpy()


        plt.figure(
            figsize=(12, 4)
        )


        plt.subplot(
            1,
            3,
            1
        )

        plt.imshow(
            original
        )

        plt.title("MRI")

        plt.axis("off")


        plt.subplot(
            1,
            3,
            2
        )

        plt.imshow(
            original
        )

        plt.imshow(
            cam,
            cmap="jet",
            alpha=0.45
        )

        plt.title(
            "Grad-CAM\n"
            f"{CLASS_NAMES[predicted_class]} "
            f"({confidence:.2f})"
        )

        plt.axis("off")


        plt.subplot(
            1,
            3,
            3
        )

        plt.imshow(
            mask,
            cmap="gray"
        )

        plt.title(
            "Tumor Mask"
        )

        plt.axis("off")


        plt.tight_layout()


        plt.savefig(
            XAI_DIR /
            f"gradcam_{index:03d}.png",
            dpi=250,
            bbox_inches="tight"
        )


        plt.show()


xai_df = pd.DataFrame(
    xai_records
)


xai_df.to_csv(
    XAI_DIR /
    "gradcam_metrics.csv",
    index=False
)


log(
    "Grad-CAM generation completed."
)


if len(xai_df) > 0:

    print(
        "\nXAI SUMMARY\n"
    )

    print(
        xai_df[
            [
                "cam_iou",
                "cam_precision",
                "cam_recall",
                "cam_area_ratio"
            ]
        ].describe()
    )


gradcam.close()


# =============================================================================
# 38 | EXPERIMENT SUMMARY
# =============================================================================

section(
    "STEP 20 | FINAL EXPERIMENT SUMMARY"
)


summary = {

    "project":
        PROJECT_NAME,

    "task":
        "3-class brain tumor MRI classification",

    "classes":
        CLASS_NAMES,

    "dataset":
        "BrainTumorDataPublic",

    "samples":
        int(len(manifest)),

    "patients":
        int(
            manifest["pid"].nunique()
        ),

    "train_samples":
        int(len(train_df)),

    "validation_samples":
        int(len(val_df)),

    "test_samples":
        int(len(test_df)),

    "train_patients":
        int(len(train_patients)),

    "validation_patients":
        int(len(val_patients)),

    "test_patients":
        int(len(test_patients)),

    "test_fold":
        TEST_FOLD,

    "model":
        "EfficientNet-B0",

    "total_parameters":
        int(total_params),

    "trainable_parameters":
        int(trainable_params),

    "image_size":
        IMAGE_SIZE,

    "batch_size":
        BATCH_SIZE,

    "epochs_configured":
        EPOCHS,

    "epochs_completed":
        len(history),

    "best_epoch":
        int(best_epoch),

    "patience":
        PATIENCE,

    "learning_rate":
        LEARNING_RATE,

    "weight_decay":
        WEIGHT_DECAY,

    "device":
        str(DEVICE),

    "gpu":
        GPU_NAME,

    "pytorch":
        torch.__version__,

    "cuda":
        torch.version.cuda,

    "seed":
        SEED,

    "best_validation_f1":
        float(best_val_f1),

    "test_metrics":
        {
            key:
                (
                    float(value)
                    if not pd.isna(value)
                    else None
                )
            for key, value
            in test_metrics.items()
        },

    "roc_auc_per_class":
        {
            key:
                (
                    float(value)
                    if not pd.isna(value)
                    else None
                )
            for key, value
            in roc_metrics.items()
        },

    "timestamp":
        datetime.now().isoformat(),

}


with open(
    METRICS_DIR /
    "experiment_summary.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )


# =============================================================================
# 39 | EXPORT MODEL
# =============================================================================

section(
    "STEP 21 | DEPLOYMENT EXPORT"
)


FINAL_MODEL_PATH = (
    EXPORT_DIR /
    "neurovision_efficientnet_b0.pth"
)


torch.save(
    model.state_dict(),
    FINAL_MODEL_PATH
)


FINAL_CHECKPOINT_PATH = (
    EXPORT_DIR /
    "neurovision_efficientnet_b0_checkpoint.pth"
)


torch.save(

    {

        "model_name":
            "EfficientNet-B0",

        "state_dict":
            model.state_dict(),

        "class_names":
            CLASS_NAMES,

        "num_classes":
            NUM_CLASSES,

        "image_size":
            IMAGE_SIZE,

        "imagenet_mean":
            IMAGENET_MEAN,

        "imagenet_std":
            IMAGENET_STD,

        "best_epoch":
            best_epoch,

        "best_validation_f1":
            best_val_f1,

        "test_metrics":
            test_metrics,

        "roc_auc_per_class":
            roc_metrics,

    },

    FINAL_CHECKPOINT_PATH
)


# =============================================================================
# 40 | DEPLOYMENT CONFIG
# =============================================================================

deployment_config = {

    "model":
        "EfficientNet-B0",

    "num_classes":
        NUM_CLASSES,

    "classes":
        CLASS_NAMES,

    "input":
        {

            "source":
                "grayscale MRI",

            "channels":
                3,

            "conversion":
                "grayscale_to_rgb",

            "size":
                [
                    IMAGE_SIZE,
                    IMAGE_SIZE
                ],

        },

    "normalization":
        {

            "mean":
                IMAGENET_MEAN,

            "std":
                IMAGENET_STD,

        },

    "mri_intensity_normalization":
        {

            "percentile_low":
                1,

            "percentile_high":
                99,

            "range":
                [
                    0,
                    255
                ],

        },

    "xai":
        {

            "method":
                "Grad-CAM",

            "target_layer":
                "model.features[-1]",

        },

}


with open(
    EXPORT_DIR /
    "config.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        deployment_config,
        file,
        indent=4
    )


# =============================================================================
# 41 | LABELS
# =============================================================================

with open(
    EXPORT_DIR /
    "labels.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        {
            str(i): name
            for i, name
            in enumerate(CLASS_NAMES)
        },
        file,
        indent=4
    )


# =============================================================================
# 42 | PREPROCESSING
# =============================================================================

preprocessing_config = {

    "input_mode":
        "grayscale",

    "convert_to_channels":
        3,

    "resize":
        [
            IMAGE_SIZE,
            IMAGE_SIZE
        ],

    "normalization":
        {

            "mean":
                IMAGENET_MEAN,

            "std":
                IMAGENET_STD,

        },

}


with open(
    EXPORT_DIR /
    "preprocessing.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        preprocessing_config,
        file,
        indent=4
    )


# =============================================================================
# 43 | DEPLOYMENT INFERENCE CODE
# =============================================================================

inference_code = r'''
import json

import numpy as np

import torch

from PIL import Image

from torchvision import models, transforms


MODEL_PATH = (
    "neurovision_efficientnet_b0.pth"
)

CONFIG_PATH = (
    "config.json"
)


with open(
    CONFIG_PATH,
    "r",
    encoding="utf-8"
) as file:

    config = json.load(file)


CLASS_NAMES = config["classes"]

IMAGE_SIZE = (
    config["input"]["size"][0]
)

MEAN = config["normalization"]["mean"]

STD = config["normalization"]["std"]


DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)


model = models.efficientnet_b0(
    weights=None
)


model.classifier[1] = (
    torch.nn.Linear(
        model.classifier[1].in_features,
        len(CLASS_NAMES)
    )
)


state_dict = torch.load(
    MODEL_PATH,
    map_location=DEVICE,
    weights_only=True
)


model.load_state_dict(
    state_dict
)


model = model.to(DEVICE)

model.eval()


transform = transforms.Compose([

    transforms.Grayscale(
        num_output_channels=3
    ),

    transforms.Resize(
        (
            IMAGE_SIZE,
            IMAGE_SIZE
        )
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=MEAN,
        std=STD
    ),

])


def predict(image):

    if not isinstance(
        image,
        Image.Image
    ):

        image = Image.fromarray(
            np.asarray(image)
        )


    image = image.convert("L")


    tensor = (
        transform(image)
        .unsqueeze(0)
        .to(DEVICE)
    )


    with torch.no_grad():

        logits = model(tensor)

        probabilities = (
            torch.softmax(
                logits,
                dim=1
            )[0]
        )


    predicted_index = int(
        probabilities.argmax()
    )


    return {

        "class":
            CLASS_NAMES[
                predicted_index
            ],

        "confidence":
            float(
                probabilities[
                    predicted_index
                ]
            ),

        "probabilities":
            {
                CLASS_NAMES[i]:
                    float(
                        probabilities[i]
                    )
                for i in range(
                    len(CLASS_NAMES)
                )
            },

    }
'''


with open(
    EXPORT_DIR /
    "inference.py",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        inference_code
    )


# =============================================================================
# 44 | REQUIREMENTS
# =============================================================================

requirements = """torch
torchvision
numpy
Pillow
gradio
huggingface_hub
"""


with open(
    EXPORT_DIR /
    "requirements.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        requirements
    )


# =============================================================================
# 45 | README
# =============================================================================

readme = f"""
# NeuroVision AI

Three-class brain tumor MRI classification.

## Classes

1. Meningioma
2. Glioma
3. Pituitary

## Dataset

BrainTumorDataPublic

Total samples: {len(manifest):,}

Unique patients: {manifest['pid'].nunique():,}

## Patient-level split

Test fold: {TEST_FOLD}

Training samples: {len(train_df):,}

Validation samples: {len(val_df):,}

Test samples: {len(test_df):,}

## Model

EfficientNet-B0

Parameters: {total_params:,}

Input: {IMAGE_SIZE} x {IMAGE_SIZE}

Input channels: 3

MRI source: grayscale

Conversion: grayscale -> RGB

## Training

Configured epochs: {EPOCHS}

Completed epochs: {len(history)}

Best epoch: {best_epoch}

Early stopping patience: {PATIENCE}

## Explainability

Grad-CAM

Target layer:

model.features[-1]

Tumor masks are used only for quantitative XAI localization evaluation.

## Best Validation F1

{best_val_f1:.4f}

## Test Metrics

"""


for key, value in test_metrics.items():

    if isinstance(
        value,
        (
            float,
            np.floating
        )
    ):

        readme += (
            f"- {key}: "
            f"{value:.4f}\n"
        )


readme += """

## ROC-AUC

Per-class ROC-AUC and macro ROC-AUC are stored in:

metrics/roc_auc_results.csv

## Export files

- neurovision_efficientnet_b0.pth
- neurovision_efficientnet_b0_checkpoint.pth
- config.json
- labels.json
- preprocessing.json
- inference.py
- requirements.txt

"""


with open(
    EXPORT_DIR /
    "README.md",
    "w",
    encoding="utf-8"
) as file:

    file.write(
        readme
    )


# =============================================================================
# 46 | FINAL FILE INVENTORY
# =============================================================================

section(
    "STEP 22 | FINAL OUTPUT INVENTORY"
)


for root, directories, files in os.walk(
    OUTPUT_ROOT
):

    for filename in files:

        path = (
            Path(root) /
            filename
        )


        size_mb = (
            path.stat().st_size /
            (1024 ** 2)
        )


        print(
            f"{path.relative_to(OUTPUT_ROOT)} "
            f"| {size_mb:.2f} MB"
        )


# =============================================================================
# 47 | FINAL SUMMARY
# =============================================================================

section(
    "NEUROVISION AI | PIPELINE COMPLETE"
)


log(
    f"Samples: "
    f"{len(manifest):,}"
)

log(
    f"Patients: "
    f"{manifest['pid'].nunique():,}"
)

log(
    f"Train: "
    f"{len(train_df):,}"
)

log(
    f"Validation: "
    f"{len(val_df):,}"
)

log(
    f"Test: "
    f"{len(test_df):,}"
)

log(
    "Model: EfficientNet-B0"
)

log(
    f"Parameters: "
    f"{total_params:,}"
)

log(
    f"Configured epochs: "
    f"{EPOCHS}"
)

log(
    f"Completed epochs: "
    f"{len(history)}"
)

log(
    f"Best epoch: "
    f"{best_epoch}"
)

log(
    f"Best validation F1: "
    f"{best_val_f1:.4f}"
)

log(
    f"Test Accuracy: "
    f"{test_metrics['accuracy']:.4f}"
)

log(
    f"Test F1 Macro: "
    f"{test_metrics['f1_macro']:.4f}"
)

log(
    f"Test Sensitivity: "
    f"{test_metrics['sensitivity_macro']:.4f}"
)

log(
    f"Test Specificity: "
    f"{test_metrics['specificity_macro']:.4f}"
)

log(
    f"Test ROC-AUC: "
    f"{test_metrics['roc_auc_ovr_macro']:.4f}"
)

log(
    f"Test PR-AUC: "
    f"{test_metrics['pr_auc_macro']:.4f}"
)

log(
    f"Macro ROC-AUC: "
    f"{macro_auc:.4f}"
)

log(
    f"Checkpoint: "
    f"{best_checkpoint}"
)

log(
    f"Deployment package: "
    f"{EXPORT_DIR}"
)

log(
    "TRAINING + EVALUATION + XAI + EXPORT COMPLETE"
)


print(
    "\n" +
    "=" * 100
)

print(
    "NEUROVISION OUTPUT:"
)

print(
    EXPORT_DIR
)

print(
    "=" * 100
)
```
